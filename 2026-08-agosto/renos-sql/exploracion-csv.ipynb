```python
import duckdb

archivo = r"dataset\reno_msisdn.csv"
envios = r"dataset\sends_msisdn.csv"

resultado = duckdb.sql(f"""
    SELECT
    *
    FROM read_csv('{archivo}')
    ORDER BY msisdn DESC;
    """)

resultado.show()
```

```python
departamento = duckdb.sql(f"""
    SELECT
    departamento,  
    count(msisdn) numero
    FROM read_csv('{archivo}')
    GROUP BY departamento
    ORDER BY numero DESC;
    """)


departamento_envios = duckdb.sql(f"""
    SELECT
    *
    FROM read_csv('{envios}')
    """)

## departamento.show()

departamento_envios.show()
```

```python
ciudades = duckdb.sql(f"""
    SELECT
    ciudad,  
    count(msisdn) numero
    FROM read_csv('{archivo}')
    GROUP BY ciudad
    ORDER BY numero DESC;
    """)

ciudades.show()
```

```python
departamento_envios_group = duckdb.sql(f"""
    SELECT
        canvas_step_nm,
        COUNT(DISTINCT msisdn) AS numero
    FROM read_csv('{envios}')
    GROUP BY
        canvas_step_nm,
    ORDER BY
        numero DESC
            """)

departamento_envios_group.show()
```

```python
import duckdb

con = duckdb.connect()

envios_path = r"C:\Users\datatigo\Desktop\tigo-sql\tigo-sql\2026-08-agosto\renos-sql\dataset\produ\envios_produ.csv"
reno_path = r"C:\Users\datatigo\Desktop\tigo-sql\tigo-sql\2026-08-agosto\renos-sql\dataset\produ\reno_produ.csv"

query = f"""
SELECT DISTINCT e.msisdn
FROM '{envios_path}' e
INNER JOIN '{reno_path}' r
    ON RIGHT(CAST(e.msisdn AS VARCHAR), 8) = RIGHT(CAST(r.msisdn AS VARCHAR), 8)
"""

resultado = con.execute(query).df()
print(resultado)

# Opcional: guardar resultado a CSV
resultado.to_csv('coincidencias.csv', index=False)
```
